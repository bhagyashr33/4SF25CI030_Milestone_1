```python
data['Date'] = data['last_review'].str.split('-').str[2]
data['Month'] = data['last_review'].str.split('-').str[1]
data['Year'] = data['last_review'].str.split('-').str[0]
data.info()
```

```python
data['Date'] = data['Date'].astype('Int64')
data['Month'] = data['Month'].astype('Int64')
data['Year'] = data['Year'].astype('Int64')
data.info()
```

```python
data = data.drop('last_review', axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder() 
```

```python
encoded_data = encoder.fit_transform(data[['neighbourhood_group','neighbourhood','room_type']]).toarray()
encoded_data
```

```python
encoded_data = pd.DataFrame(encoder.fit_transform(data[['neighbourhood_group','neighbourhood','room_type']]).toarray(),columns=encoder.get_feature_names_out(),index=data.index)
encoded_data
```

```python
data = data.drop(columns=['neighbourhood_group', 'neighbourhood', 'room_type'])
```

```python
data = pd.concat([data, encoded_data], axis=1)
data.head()
```
